```python
from bs4 import BeautifulSoup
import requests
import pandas as pd

response = requests.get("https://appbrewery.github.io/news.ycombinator.com/")
response.raise_for_status()
yc_web_page = response.text

soup = BeautifulSoup(yc_web_page, "html.parser")

comment_tags = soup.find_all("a", string=lambda text: "comments" in text if text else False)

comments_data = []
for tag in comment_tags:
    comment_text = tag.getText()
    comment_link = tag.get("href")
    comments_data.append({"Comments": comment_text, "Link": comment_link})

df = pd.DataFrame(comments_data)
output_file = "comments_data.xlsx"
df.to_excel(output_file, index=False)

print(f"Comments data saved to {output_file}")
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

url = "https://www.udemy.com/course/100-days-of-code/learn/lecture/21733858#overview"

headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/58.0.3029.110 Safari/537.36'
}

response = requests.get(url, headers=headers)

soup = BeautifulSoup(response.text, 'html.parser')

span_tag = soup.find('span', class_='truncate-with-tooltip--ellipsis--YJw4N')

if span_tag:
    text_content = span_tag.text.strip()
    print("Text content:", text_content)
else:
    text_content = "Element not found!"
    print(text_content)

data = {'Scraped Text': [text_content]}

df = pd.DataFrame(data)

excel_file = "scraped_content.xlsx"

df.to_excel(excel_file, index=False)

print(f"Scraped content saved to {excel_file}")
```

```python
from bs4 import BeautifulSoup
import requests
import pandas as pd

response = requests.get("https://web.archive.org/web/20200518073855/https://www.empireonline.com/movies/features/best-movies-2/")
yc_web_page = response.text

soup = BeautifulSoup(yc_web_page, "html.parser")

movie_name = soup.find_all("h3", class_="title")
article_texts1 = [(tag.getText().split()[0]) for tag in movie_name]
article_texts2 = [' '.join(tag.getText().split()[1:]) for tag in movie_name]

article_texts1 = article_texts1[::-1]
article_texts2 = article_texts2[::-1]

data={"number":article_texts1,"name":article_texts2}
df = pd.DataFrame(data)
excel_file = "movie_list3.xlsx"
df.to_excel(excel_file, index=False)
```
